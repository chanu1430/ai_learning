# DATA CLEANING

```python
import numpy as np
import pandas as pd
```

```python
data = {
    "age":[25,np.nan,30,35,40,1000],
    "salary":[40000,50000,np.nan,70000,80000,9999999],
    "gender":["M","F",np.nan,"M","M","F"]
}
```

```python
df= pd.DataFrame(data)
# df.head()
df
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
# df=df.dropna()
# df

df["age"]=df["age"].fillna((df["age"].median()))
df["salary"]=df["salary"].fillna((df["salary"]).mean())
df["gender"]=df["gender"].fillna((df["gender"]).mode()[0])
df
```

```python
q1=df["salary"].quantile(0.25)
q3=df["salary"].quantile(0.75)
iqr=q3-q1
iqr
```

```python
lower = q1 - 1.5 * iqr

upper = q3 + 1.5 * iqr
print(f"lower : {lower}")
print(f"upper: {upper}")
```

```python
outliers = df[
    (df["salary"] < lower)
    |
    (df["salary"] > upper)
]
outliers
```

```python
df=df[(df["salary"]>=lower) & (df["salary"]<=upper)]
df
```
