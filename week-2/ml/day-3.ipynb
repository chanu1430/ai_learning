# FEATURE ENGINEERING

```python
import pandas as pd
```

```python
data = {
    "salary":[40000,50000,70000,90000,120000],
    "experience":[1,2,4,6,10],
    "height":[170,180,175,165,172],
    "weight":[70,85,75,60,78]
}
```

```python
df=pd.DataFrame(data)
df
```

```python
df["height(m)"]=df["height"]/100
df["bmi"]=df["weight"]/(df["height(m)"]**2)
df
```

```python
df["salary_per_experience"]=df["salary"]/df["experience"]
df
```

```python
df["rank"]=pd.cut(
    df["experience"],
    bins=[0,3,5,7,10],
    labels=["junior","assosiate","senior","lead"]

)
df
```

```python
print(df.corr(numeric_only=True))
```
